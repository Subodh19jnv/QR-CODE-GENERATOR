```python
import qrcode
from IPython.display import display

print("QR CODE GENERATOR")

data = input("Enter text or website URL: ")

color = input("Enter QR color: ").strip()
border = int(input("Enter border size: "))

qr = qrcode.QRCode(
    box_size=10,
    border=border
)

qr.add_data(data)
qr.make(fit=True)

img = qr.make_image(
    fill_color=color,
    back_color="white"
)

display(img)

img.save("my_qr.png")

print("QR Code generated successfully!")
print("Image saved as my_qr.png")
```
